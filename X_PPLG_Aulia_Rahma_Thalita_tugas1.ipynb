```python
%%writefile praktik_variabel.cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    // Deklarasi Variabel dan Tipe Data
    string namaBarang = "iPhone XS";
    int jumlah = 2;
    float harga = 7500000.50;
    char kategori = 'A';
    bool tersedia = true;

    // Menampilkan Output ke Layar
    cout << "=== DATA INVENTARIS ===" << endl;
    cout << "iPhone XS: " << namaBarang << endl;
    cout << "Kategori    : " << kategori << endl;
    cout << "Jumlah      : " << jumlah << endl;
    cout << "75000000.50 : Rp " << harga << endl;
    cout << "Tersedia    : " << tersedia << " (1 = Ya, 0 = Tidak)" << endl;
    cout << "=======================" << endl;

    return 0;
}
```
